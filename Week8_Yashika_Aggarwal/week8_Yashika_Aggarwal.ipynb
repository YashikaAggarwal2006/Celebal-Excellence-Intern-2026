# 🧠 Single Agent Pipeline Project

## Problem Statement
Build a **Single-Agent Smart Assistant** that:
- Understands user queries
- Routes tasks based on intent
- Uses tools when required
- Returns structured JSON output

### The agent should handle:
- Math queries → Calculator Tool
- Keyword extraction → Keyword Tool
- General queries → Direct response

---
### 🛠️ What You Need to Implement
- Agent logic
- Conditional routing
- Tool integration
- Basic error handling

### 🚀 Bonus
- Improve routing
- Add logging

````python
import re

def calculator(expression: str) -> str:
    """Evaluate a mathematical expression."""
    try:
        if not re.fullmatch(r"[0-9\.\+\-\*/\(\)\s]+", expression):
            return "Error in calculation"
        return str(eval(expression))
    except Exception:
        return "Error in calculation"
````

````python
# 🛠️ TOOL 2: Keyword Extractor

def extract_keywords(text: str) -> list:
    """Extract keywords from text."""
    try:
        words = text.split()
        keywords = list(set([w.lower().strip(".,!?") for w in words if len(w) > 4]))
        return keywords[:5]
    except Exception:
        return []
````

````python
import logging

logging.basicConfig(level=logging.INFO, format="%(asctime)s [%(levelname)s] %(message)s")
logger = logging.getLogger("agent")
````

## 🤖 Implement Agent Logic Below

👉 Use conditional routing:
- If query contains "calculate" → use calculator
- If query contains "keywords" → use keyword extractor
- Else → general response

````python
# 🤖 AGENT FUNCTION (TO IMPLEMENT)

def agent(query: str):
    logger.info(f"Received query: {query}")
    query_lower = query.lower()

    try:
        if "calculate" in query_lower:
            match = re.search(r"calculate\s+(.*)", query, re.IGNORECASE)
            expression = match.group(1).strip() if match else ""

            if not expression:
                logger.warning("No expression found after 'calculate'")
                return {"type": "error", "result": "No expression provided to calculate."}

            result = calculator(expression)
            if result == "Error in calculation":
                logger.error(f"Calculation failed for expression: {expression}")
                return {"type": "error", "result": result}

            logger.info(f"Calculation success: {expression} = {result}")
            return {"type": "calculation", "result": result}

        elif "keywords" in query_lower:
            match = re.search(r"keywords\s*(?:from|of|in)?\s*(.*)", query, re.IGNORECASE)
            text = match.group(1).strip() if match else query

            if not text:
                logger.warning("No text found for keyword extraction")
                return {"type": "error", "result": "No text provided for keyword extraction."}

            result = extract_keywords(text)
            logger.info(f"Extracted keywords: {result}")
            return {"type": "keywords", "result": result}

        else:
            logger.info("Routed to general response")
            return {"type": "general", "result": f"You asked: '{query}'. I don't have a specific tool for this, but I registered your query."}

    except Exception as e:
        logger.exception("Unexpected error in agent")
        return {"type": "error", "result": f"Unexpected error: {str(e)}"}
````

## 📦 Expected Output Format

```
{
  "type": "calculation / keywords / general / error",
  "result": ...
}
```

````python
# 🧪 Test Cases

queries = [
    "Calculate 20 + 5",
    "Extract keywords from Artificial Intelligence is transforming industries",
    "What is machine learning?",
    "Calculate 10 / 0",
    "Calculate import os",
]

for q in queries:
    print("Query:", q)
    print("Response:", agent(q))
    print("-" * 50)
````

````python
# 🎯 Interactive Mode

while True:
    user_input = input("Enter query (type 'exit' to stop): ")
    if user_input.lower() == "exit":
        break
    print("Response:", agent(user_input))
````

## ✅ Conclusion

Built a single-agent assistant that routes queries by intent:
- **"calculate"** → Calculator Tool
- **"keywords"** → Keyword Extractor Tool
- anything else → General response

All responses follow a structured `{"type": ..., "result": ...}` format, with error handling for invalid inputs and logging for traceability.
